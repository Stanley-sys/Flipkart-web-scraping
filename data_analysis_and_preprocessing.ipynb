```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv("Flipkart_mobile_phones.csv")
df
```

```python
df.drop("Unnamed: 0",axis=1,inplace=True)
```

```python
df
```

```python
sale_price=[]
actual_price=[]
for i in range(df.shape[0]):
    try:
        a=df["Sale & Actual Price"][i].split(df["Offer"][i])[0].split("₹")[1]
        sale_price.append(int(a.replace(",","")))
        try:
           b=df["Sale & Actual Price"][i].split(df["Offer"][i])[0].split("₹")[2]
        except IndexError:
           b=a
        actual_price.append(int(b.replace(",","")))
    except IndexError:
        sale_price.append(df["Sale & Actual Price"][i])
        actual_price.append(df["Sale & Actual Price"][i])
```

```python
df["Sale Price"]=sale_price
df["Actual Price"]=actual_price
```

```python
df.drop("Sale & Actual Price",axis=1,inplace=True)
```

```python
df
```

```python
splitted_cols=df["Review & Ratings"].str.split("&",expand=True)
```

```python
df["No of Reviews"]=splitted_cols[0].str.replace(",","").str.extract(r"(\d+)").astype('Int64')
df["No of Ratings"]=splitted_cols[1].str.replace(",","").str.extract(r"(\d+)").astype('Int64')
```

```python
df.drop("Review & Ratings",axis=1,inplace=True)
```

```python
df
```

```python
df.to_csv("Proccessed_Flipkart_mobile_phones.csv")
```
